# Introduction to Exploratory Analysis

It is essential to load, explore, and understand data, its quality, and its behavior **before** beginning to work with it. This process is necessary to gain an initial approximation of the data and must be done prior to conducting any analysis. Furthermore, we must understand data behavior because **many statistical data analysis techniques presuppose the fulfillment of certain preconditions**, such as **normality**, **non-collinearity**, and **comparable scales**, in order to guarantee that the techniques function correctly.

The formal definition of this exploratory data analysis is: the set of statistical techniques whose objective is to explore, describe, and summarize the nature of the data and understand the relationships existing between the variables of interest, maximizing comprehension of the dataset.

## Why We Must Use Exploratory Analysis:

Identify potential errors (incorrectly entered data, missing values, or poor variable encoding), reveal the presence of outliers, check the relationships between variables (correlations) and their potential redundancy, and perform descriptive data analysis using graphical representations and summaries of the most significant aspects. In my opinion, [this]() is a good article on exploratory analysis, in case you want to dive deeper. In it, the author explains that the goal of EDA can be divided into 6 principles (which I have modified slightly according to my judgment):

* **Maximize the number of insights we get from the dataset**
* **Discover the structure and relationships among the data**
* **Extract relevant variables**
* **Detect anomalies, duplicate values, and missing values**
* **Validate assumptions or presuppositions we may have about the data**
* **Determine the relevance of the data and optimize the input for our models**

An important concept here is that EDA is not a process exclusive to data science modeling. People often talk about the **80-20% rule**, where 80% of a Data Scientist's work should be dedicated to obtaining and understanding the data, and 20% to modeling and model tuning tasks. ## Steps of Exploratory Analysis

There are hundreds of tutorials and ways to approach an exploratory analysis. **The steps included below are based on my personal experience**, but they must be **adjusted slightly on a case-by-case basis** because every problem is different. We will see that exploratory analysis can be divided into **3 major steps**, but these steps are subdivided into many tasks that we must perform to start interpreting the results. An **important note** about EDA is that we will look at data cleaning first and then visualization, but it is very important to highlight that in practice we do both at the same time, since we often need to visualize the data to identify null values or errors.

<br>

* **Data Loading:** This step seems trivial but it is not in practice, and unfortunately it is hard to see during course exercises, where we simply have to read the data. Accessing data in practice generally involves connecting to a database, a data lake, or a cloud platform, or even combining data from different sources. It is important **not to underestimate** this step, since we will not always have the data readily available. There is something very important, especially when reading a text file, which is known as **encoding**. Have you ever read an Excel file and gotten weird characters? This happens because the encoding Excel is using to read the file is not the correct one. This is often a problem when dealing with text read from text files.
* **Data Cleaning:** Data cleaning is based on obtaining a dataset with which we can visualize and ultimately model the problem. In this step, we will try to remove duplicates (if applicable), handle null values (if applicable), and ensure that data formats and types are correct. I have also added a bonus topic for when we deal with string-type data: encoding.
* **Formats and Types:** One of the first tasks we must check when facing a new dataset is the data types we have. We can do this with the `.info()` method or even the `.dtypes` method on the DataFrame. What we want is to familiarize ourselves with the data types we have, and also ensure the format is correct. In geographical datasets, for example, we must verify that coordinate columns are in the correct format.
* **Null Values:** One of the most important jobs in exploratory analysis happens with null values. Null values can occur due to poor data collection, corrupted files, or they can be intentional. We will see that there are many different ways to solve the problem of null values, but we must always keep in mind that we are often manufacturing information if we choose to impute a value to these blanks, so we must be cautious. The [.isnull() or .isna()](https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.isnull.html) method applied to DataFrame columns will be very useful in these cases. We can do `df[column].isna().sum()` to quickly find out how many null elements we have per column.
* **Duplicate Records:** Duplicate records are not always dealt with, but they can be annoying from time to time, although we must be sure of their origin before deciding whether to remove them or not. Basically, a duplicate record introduces a problem to ML systems that may be desirable or not: the magnification of errors. If we have a repeated record 10 times and our model makes a prediction error, it will be making that same error 10 times for a single record. This can be interesting in some cases, for example when there are few instances of the positive class, we often do "oversampling" to create this exact effect in the model.

<br>

* **Visualization:** This is the stage people normally refer to as EDA. From this point on, we have a clean and functional dataset. Now it's just a matter of exploring it, visualizing the distribution of numerical variables, exploring the counts of categorical variables, applying statistical analysis, and running transformations. At the end of this process, we should have our data ready to start modeling. This process has the least structure, because as we start exploring the data, we will discover new avenues to explore, and it is hard to anticipate that without seeing the dataset. But there are 4 tasks that we will practically always do (and depending on your level of formality, even 5):
* **Split:** This is a key task in any Machine Learning model we build. ML models are trained with data; for example, if we want to train a classification model, our model will try to create a separation boundary (which does not have to be linear) between the two classes. We need this separation to work well with **any type of data** coming from the same nature.
* But how do we guarantee that the model works well and, more importantly, that it **generalizes** well? What we do is divide our data into training data and testing data (and sometimes training, validation, and testing data). During model training, we set aside a dataset that the model does not see, and then we evaluate the model with that dataset to see if it generalizes well what it learned during training.
* Why is it important to do a split now? Because we want to prevent **Data Leakage**, even in visualizations, so as not to bias ourselves toward the resource we will use to evaluate the model. We want the model to never receive information about the test set. Below we see a typical division into training, validation, and test sets.
* **Univariate Analysis:** We must explore all variables by themselves. How are they distributed? Which ones are relevant? Is there any peculiarity that escaped us during the previous two processes? If the variables are categorical... what are the counts per category? Is any category more frequent than the others? Less frequent?
* **Bivariate Analysis:** This step adds one more dimension to the problem. Now we will try to study dependency relationships between variables in pairs. Something important to keep in mind is that we should always try to visualize what relationship our variables have with the target variable we want to predict. We should also look for relationships among the other variables to ensure there are no linear dependencies between them that could reduce model performance.
* **Multivariate Analysis:** In some cases, we can try to visualize relationships among multiple variables simultaneously. These become more complex to interpret, but in some cases, they can give us very useful information to solve our problem. Remember that our goal in this step is to identify the subset of variables that will yield the best predictive result.
* **Correlation Analysis:** In correlation analysis, we will study correlation relationships between continuous variables. If the variable we want to predict is continuous, then we are interested in including it in the analysis and are particularly interested in variables that correlate well with our target variable. Otherwise, correlation is still useful for eliminating variables that have linear dependencies among themselves.

<br>

* **Other Analyses:** Here we will study other processes that could be interesting for our study, such as feature engineering, dimensionality reduction, and data augmentation. In this stage, we will also apply transformations to the data. To verify that transformations have been successful, we can perform statistical tests such as normality tests.
* **Feature engineering** is the process by which we "manufacture" new variables from the data we have. For example, if we have an age variable and want to predict customer churn, perhaps by doing the analysis we will see that most churning customers are under 20 or over 60. Therefore, instead of having a numeric age variable, we could convert it into categorical bins such as "under 20", "between 20 and 60", and "over 60". This model will likely perform even better because it captures changes relative to the target variable more effectively.
* **Dimensionality reduction** is the opposite case of feature engineering. In ML we must be very cautious with extremes; we do not want irrelevant variables (which is why we use feature engineering to refine their use for prediction). But we also do not want so many variables that our model gets lost in the dimensionality of the problem. This problem is so common it has a name: "The Curse of Dimensionality." If we think of datasets as tables, a high-dimensionality problem is one with many columns. One tool to solve that problem is dimensionality reduction. Basically, we will look for statistical methods to convert those $N$ variables into $M$, where $M \ll N$.
* **Data augmentation** is another form of feature engineering, where we incorporate more data into our dataset, either by adding new variables (such as weather variables) or by modifying existing variables to enrich the dataset (this is generally applied in Deep Learning).
* **Transformations** will be another step in this process. As we have already mentioned, we often want to transform our variables to another scale or convert a heavy-tailed distribution into a distribution closer to a normal distribution. We can perform both processes using transformations and rescaling, as appropriate. For a review on this, check out the notebook on [Python visualization](https://www.kaggle.com/code/micheldc55/tutorial-data-science-matplotlib-y-seaborn).

## Correlation Analysis

This analysis is always very important for numerical variables, especially in linear models (though we should always check it). Multicollinearity is a major problem in matrix computation, so we must avoid variables with high correlation at all costs, because it means they are introducing redundant noise into the model we will later train.

Some common mistakes made in correlation analysis include:

* **Including categorical variables.**
* **Not defining the correct scale:** Correlation ranges between -1 and 1, but often when plotting, the color scale automatically adjusts to the minimum and maximum values present in our correlation matrix rather than the fixed global limits. This can visually bias us into thinking a difference between two values is greater than it actually is. To fix this, we typically enforce the color scale limits (`vmin=-1`, `vmax=1`) on the heatmap.
* **Judging correlation as "good" or "bad" based on an arbitrary number:** We often see rules of thumb like "correlation above 0.7 is very high." In reality, we must evaluate both the scatter plots and the numerical correlation values together to make an informed decision.
* **Not annotating the heatmap values:** Displaying the numerical correlation coefficients directly on the cells helps resolve ambiguity when we aren't sure if what we see is a very high or moderate correlation. We achieve this by passing the `annot=True` parameter to Seaborn's heatmap.

````python
df_train_2 = df_train.drop(columns=['species'])


sns.heatmap(df_train_2.corr(), vmin=-1, vmax=1, cmap='magma', annot=True)
````

It is clearly seen that the correlation between `petal_length` and `sepal_length`, as well as between `petal_length` and `petal_width`, is high, as we saw in the other analyses. This confirms our theory. Let's see what happens to the scale if we don't specify the `vmax` and `vmin` values...

````python
sns.heatmap(df_train.corr(numeric_only=True), cmap='magma', annot=True)
plt.show()
````

We see that the colors are more intense because the scale is narrower, so a small change significantly affects the perception of the difference between the correlation values.

## Introducing a Model

We have not yet resolved the numerical values issue. The target class we want to predict has text values, and although some ML models allow this, it is generally ideal to convert all variables to numerical values to take advantage of NumPy's vectorized computation. Below, we quickly convert the values of the 3 classes into numbers.

````python
df_train['species'].unique()
````

````python
map_target_class = {'Iris-setosa': 0, 'Iris-versicolor': 1, 'Iris-virginica': 2}

df_train['species_numerical'] = df_train['species'].map(map_target_class)
df_test['species_numerical'] = df_test['species'].map(map_target_class)

df_train.head(10)
````

````python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report, confusion_matrix
from sklearn.preprocessing import StandardScaler
````

````python
X_train = df_train.drop(['species', 'species_numerical'], axis=1)
y_train = df_train['species_numerical']
X_test  = df_test.drop(['species', 'species_numerical'], axis=1)
y_test  = df_test['species_numerical']
````

````python
scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test  = scaler.transform(X_test)
````

````python
lr = LogisticRegression(random_state=101, max_iter=1000)
lr.fit(X_train, y_train)
````

````python
preds_train = lr.predict(X_train)

# Create the confusion matrix
cm_train = confusion_matrix(y_train, preds_train)

# Plot the confusion matrix as a heatmap
plt.figure(figsize=(8, 6))
sns.heatmap(cm_train, annot=True, fmt='d', cmap='Blues', cbar=False,
            xticklabels=df_train['species'].unique(), yticklabels=df_train['species'].unique())
plt.xlabel('Predicted Label')
plt.ylabel('True Label')
plt.title('Confusion Matrix - Training Data')
plt.show()

print(classification_report(y_train, preds_train))
````

We see that the accuracy of the result is **extremely high**. This is reasonable because we had already seen that it is possible to obtain a fairly good separation boundary just by looking at the problem, so it was quite likely that our algorithm (even though we don't know how it works yet) would be able to return a very good result.

The question we must ask ourselves is... **Are we done?**

We could look at that result and think yes, we have achieved a very high accuracy in our model, it's ready to deploy! The truth is no, because we have only validated with the set we trained the model on. Later on we will see the risks of overfitting, but the important thing is to understand that we haven't measured how our model performs on new data!

Let's see how we can do that below...

````python
preds = lr.predict(X_test)

# Create the confusion matrix for test data
cm_test = confusion_matrix(y_test, preds)

# Plot the confusion matrix as a heatmap
plt.figure(figsize=(8, 6))
sns.heatmap(cm_test, annot=True, fmt='d', cmap='Blues', cbar=False,
            xticklabels=df_train['species'].unique(), yticklabels=df_train['species'].unique())
plt.xlabel('Predicted Label')
plt.ylabel('True Label')
plt.title('Confusion Matrix - Test Data')
plt.show()

print(classification_report(y_test, preds))
````

Officially, we see that our model is just as good with data it has never seen before. Therefore, we can say that the model appears to generalize well on our dataset. It is important to point out that our total sample size is quite small, so in a real-world scenario, we would certainly need to test on other datasets or try to acquire more data to provide a more confident assessment.

## How to Handle Null Values

The handling of null values is often the subject of many online tutorials, but it is important to clarify a couple of details about this process. We have two main ways to deal with null values:

* **Dropping records:** In this case, we assume that the information in the nulls is not relevant to the problem or we cannot provide useful information to the model if that value is null, so we remove it. We can do this at the row level (dropping rows that contain nulls) or at the column level if we see that a column has a disproportionate amount of null values. This method has a major disadvantage: we lose an entire row of records or an entire column (we are losing information). But it has a big advantage: we are not inventing values for the records.
* **Imputing records:** In this case, we do the opposite; we assume we do not want to lose the information contained within the records that have null values, and we decide to **impute** a value for the null, using one of the criteria we will see below.
* **Imputing with a measure of central tendency:** The most common is the mean or the median. Basically, we can impute the mean value of the column to all null values in that column. We could even group by categories and refine the mean value further by category for imputation. If you want to see a practical example of this, you can check it [here](https://www.kaggle.com/code/micheldc55/tutorial-data-science-matplotlib-y-seaborn).
* **Imputing with an "unknown" category:** Often, when we have nulls in categorical variables, a good option is to impute a "new category" for the nulls, which could be "unknown", "desconocido", etc. The advantage of this is that we are not inventing values, but the disadvantage is that we are increasing the dimensionality of the problem because we are adding a new value whose behavior the model will have to learn.
* **Imputing a known value:** This is not very frequent in Kaggle problems because we do not have real-world knowledge of the problem, but in practice, it is possible that we can solve the issue if we understand why that null is generated. Imagine that when users enter 0 in a variable, due to some internal glitch the system fails to process this value and returns a null. If we knew this, we could fix it and also recover the actual value that should go in that position.
* **Imputing with an ML model:** This sounds a bit like solving Machine Learning with Machine Learning, and it sort of is. In this case, suppose we have columns where we detect nulls. What we do is build a model that tries to predict the value of that null based on the values of the other records that do not have a null in that position. **Careful!** In this process, we must not use the target variable, as we could incur data leakage if we later use the predicted values for the null to predict the target. Scikit-learn has a model within the **impute** module for this, called **IterativeImputer**. They also provide a class when we want to fill nulls with a value like the median or a constant, called SimpleImputer, though for the latter I prefer to use the pandas `.fillna()` method.



```python
from sklearn.impute import IterativeImputer
from sklearn.impute SimpleImputer

imputer = IterativeImputer()
imputed_df = imputer.fit_transform(df)

```

And while I agree that this is a critically important process, I also feel it is very important to clarify that whenever we impute null values, we are **inventing values**. Obviously, we are inventing values by making an informed decision using data, but generally—unless the creator of the dataset tells us directly—we will never know if those null values are imputed correctly or not. Therefore, it is a process that must be done with **great care**.

## Special Case of Dimensionality Reduction: PCA

### What is Dimensionality Reduction?

We often think that the more information we give to the model, the better it will perform, but this line of thought is usually problematic.

Generally, we will find that we try to reduce the dimensionality of the problem primarily to solve two issues: variable separability and/or visualization, and computation. When we reduce the dimensionality of the problem, we can narrow it down to 2 or 3 variables and thus be able to **visualize** the problem. This is key, because we will often see that a multi-dimensional problem can be reduced to 2 or 3 dimensions while retaining a large portion of the class separation. This allows us to visually understand our entire problem as a representation of 2 or 3 variables instead of 10 (for example).

The other issue is computing power, both for model training and production deployment. If we have a model with 20 features to predict a target, the model trains by receiving 20-element instances and performing the corresponding operations depending on the model type. This implies that the model must train in $N \times m$ dimensions, where $N$ is the number of records and $m$ is the number of variables (in this case 10). If we use dimensionality reduction techniques and bring the problem down to $N \times m'$ dimensions, where $m'$ is the reduced number of variables (generally 2 or at most 3), we have reduced the volume of computations the model must perform proportionally to $m'/m$. This makes our problem much more efficient in terms of computational cost (requiring less processing power to train).

In this process, irrelevant (or marginally relevant) variables are identified and removed, reducing the noise of the original dataset. The best model we can achieve is not the one that takes the most variables into account, but the one that includes the highest amount of quality information.

### Dimensionality Reduction Techniques:

Feature selection: filtering low-variance variables, univariate statistical tests (dependency tests), recursive feature elimination, random forests, missing value ratio, filtering highly correlated variables, etc.

#### Mathematical Dimensionality Reduction Techniques:

Despite being able to reduce dimensionality by dropping or adding columns, often our variables provide little information, but we are still interested in including them. For this, there are also techniques that allow us to take our feature space and convert it into a subspace of components that attempt to preserve some property of the initial space. This could be variance, point-to-point distance, etc. If you are interested, [here](https://hmong.es/wiki/Nonlinear_dimensionality_reduction) most of the models are summarized (and in Spanish).

* Linear: factor analysis, principal component analysis (PCA), linear discriminant analysis.
* Non-linear: multidimensional scaling, t-SNE, isometric feature mapping (ISOMAP), uniform manifold approximation and projection (UMAP).

Let's see what the application of a linear method like PCA would look like for this case. We have 4 variables, so we cannot directly visualize all 4 variables to identify possible patterns among them. We will try to use the Principal Component Analysis technique to obtain a projection into 2 components that attempt to capture most of the variance of the 4 initial variables.

PCA uses the eigenvalues of the matrix we pass as input, and creates its eigendecomposition into eigenvectors. When we select the number of variables we want to consider (`n_components=2` in this case), we are telling the model that we want to keep the first two eigenvectors and project our data (from 4 dimensions) onto these two directions.
